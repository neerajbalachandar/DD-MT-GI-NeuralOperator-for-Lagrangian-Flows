```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("output/benchmark_task1/particle_state_error_timeseries.csv")
```

```python
plt.figure(figsize=(9,6))

plt.semilogy(df["time"], df["position_rel_l2"], lw=2, label=r"$\vec{x}$")
plt.semilogy(df["time"], df["gamma_rel_l2"], lw=2, label=r"$\vec{\Gamma}$")
plt.semilogy(df["time"], df["sigma_rel_l2"], lw=2, label=r"$\sigma$")

plt.xlabel("Simulation Time [s]", fontsize=12)
plt.ylabel("Relative L2 Error", fontsize=12)
plt.xlim(0,0.65)
# plt.grid(True, which="both", alpha=0.3)
plt.grid(False)
plt.legend()
plt.tight_layout()
plt.savefig("output/benchmark_task1/relative_l2_error.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(df["time"], df["position_rmse"], lw=2, label="Position")
plt.plot(df["time"], df["gamma_rmse"], lw=2, label=r"$\Gamma$")
plt.plot(df["time"], df["sigma_rmse"], lw=2, label=r"$\sigma$")

plt.xlabel("Simulation Time [s]")
plt.ylabel("RMSE")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("rmse.png", dpi=300)
plt.show()
```
