# Analyze GINO Evaluation Results

Review per-pair scores and native field evaluation grouped by test scenario.

```python
from pathlib import Path
import json
import pandas as pd
RESULTS = Path('../runs/baseline/evaluation')
one_step = pd.DataFrame(json.loads((RESULTS / 'one_step.json').read_text())['records'])
print('one-step records:', len(one_step))
one_step.describe(include='all')
```

```python
category_path = RESULTS / 'native_task2_by_category.json'
if category_path.exists():
    by_category = json.loads(category_path.read_text())
    summary = pd.DataFrame([{'category': key, 'frames': value['n_frames'], 'cases': ', '.join(value['cases']), 'mean_velocity_mse': value['mean_velocity_mse'], 'mean_velocity_relative_l2': value['mean_velocity_relative_l2']} for key, value in by_category.items()])
    display(summary.sort_values('category'))
```

```python
if (RESULTS / 'rollout.csv').exists():
    rollout = pd.read_csv(RESULTS / 'rollout.csv')
    display(rollout.groupby('horizon')[['mse', 'relative_l2']].agg(['mean', 'median']))
```
