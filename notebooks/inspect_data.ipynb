# Inspect Generated VPM Data

Inspect pair counts, case splits, test scenario labels, and the particle and field resolutions in the processed archive.

```python
from pathlib import Path
import sys, numpy as np
ROOT = Path.cwd()
if not (ROOT / 'gino').is_dir(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from gino.data.dataset import load_processed_dataset
DATA = ROOT / 'processed_data_vpm/particle_evolution_dataset.npz'
data = load_processed_dataset(DATA)
print('features:', data['feature_names'].tolist())
print('pairs train/val/test:', *(len(data[k]) for k in ('train_pair_ids','val_pair_ids','test_pair_ids')))
print('cases train/val/test:', *(len(data[k]) for k in ('train_cases','val_cases','test_cases')))
print('state order:', data['state_names'].tolist())
```

```python
contexts = data['pair_contexts']
case_to_categories = {}
for pair_id in data['test_pair_ids']:
    context = contexts[int(pair_id)]
    if isinstance(context, np.ndarray): context = context.item()
    case_to_categories.setdefault(str(context.get('case')), tuple(context.get('evaluation_categories', ())))
for case, categories in sorted(case_to_categories.items()): print(f"{case}: {', '.join(categories)}")
```

```python
counts = {}
for pair_id in data['test_pair_ids']:
    context = contexts[int(pair_id)]
    if isinstance(context, np.ndarray): context = context.item()
    for category in context.get('evaluation_categories', []): counts[category] = counts.get(category, 0) + 1
print('test pairs by category:', counts)
```
