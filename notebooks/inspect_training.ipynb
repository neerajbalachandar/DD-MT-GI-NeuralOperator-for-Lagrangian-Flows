# Inspect Training

Load saved epoch metrics and checkpoint metadata. The trainer remains the implementation source of truth.

```python
from pathlib import Path
import json, pandas as pd
RUN = Path('../runs/baseline')
history = pd.DataFrame(json.loads((RUN / 'history.json').read_text()))
display(history.tail())
checkpoint_metadata = __import__('torch').load(RUN / 'best_model.pt', map_location='cpu', weights_only=False)
checkpoint_metadata.keys()
```

```python
checkpoint = checkpoint_metadata
meta = checkpoint.get('metadata', {})
split = meta.get('split', {})
print('dataset:', meta.get('dataset_path'))
print('pair counts train/val/test:', *(len(split.get(key, [])) for key in ('train_pair_ids', 'val_pair_ids', 'test_pair_ids')))
print('test scenarios:')
for case, categories in sorted(split.get('test_case_categories', {}).items()): print(f'{case}: {categories}')
```
