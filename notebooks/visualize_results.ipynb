# Visualize Saved Evaluation Results

Compare native field errors across test scenarios and snapshots.

```python
from pathlib import Path
import json
import matplotlib.pyplot as plt
import pandas as pd
RESULTS = Path('../runs/baseline/evaluation')
category_file = RESULTS / 'native_task2_by_category.json'
if category_file.exists():
    categories = json.loads(category_file.read_text())
    summary = pd.DataFrame([{'category': name, 'relative_l2': row['mean_velocity_relative_l2']} for name, row in categories.items()])
    ax = summary.sort_values('category').plot.bar(x='category', y='relative_l2', legend=False, rot=35)
    ax.set_ylabel('Mean native velocity relative L2')
    plt.tight_layout(); plt.show()
```

```python
one_step = json.loads((RESULTS / 'one_step.json').read_text())['records']
plt.plot([r['pair_id'] for r in one_step], [r['mse'] for r in one_step])
plt.xlabel('Test pair'); plt.ylabel('Particle state MSE'); plt.show()
```
